## 1. Load the Data

Read the raw CSV and confirm the number of rows and columns.

```python
import pandas as pd

df = pd.read_csv("data/diabetes_risk.csv")
```

## 3. Column Types and Non-Null Counts

Use `df.info()` to see each column's data type and how many non-null values it has. Columns with fewer than 15,000 non-null entries have missing values.

**What to look for:**
- Are numeric columns stored as numbers (int64 / float64)?
- Are categorical columns stored as strings?
- Which columns have missing data?

```python
df.info()
```

**Observations:**
- All data types look correct — numerics are int64/float64, categoricals are strings.
- Three columns have missing values: `smoking_status` (14,528/15,000), `alcohol_consumption` (11,212/15,000), and `income_bracket` (14,537/15,000).
- `patient_id` is a row identifier, not a feature — it will be excluded during modelling.

## 4. Summary Statistics (Numeric Columns)

Review the central tendency, spread, and range of each numeric feature.

**What to look for:**
- Unrealistic min/max values (e.g. negative ages, BMI of 0)
- Large standard deviations suggesting high variability
- Skewness (big gap between mean and 50th percentile)

```python
df.describe()
```

---
## 5. Missing Values

Count the exact number and percentage of missing values per column. This tells us what needs to be handled during data cleaning.

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(1)
missing_df = pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
missing_df[missing_df['missing_count'] > 0]
```

**Observations:**
- `alcohol_consumption` has the most missing values (~25.3%) — too many rows to simply drop.
- `smoking_status` and `income_bracket` are each ~3.1% missing.
- All three are categorical columns. The decision on how to handle them is documented in `data-cleaning.ipynb`.

---
## 6. Target Variable Distribution

Check the distribution of `diabetes_risk` (our target). An imbalanced target will affect model selection and evaluation strategy.

```python
df['diabetes_risk'].value_counts()
```

**Observations:**
- The target is **imbalanced**: Low (60%), Moderate (25%), High (15%).
- The "High" class is the most important for the clinic (patients to prioritize for screening) but is also the smallest group.
- This imbalance will need to be addressed during modelling (e.g. class weights, stratified sampling, recall-focused evaluation).

---
## 7. Duplicate Check

Verify there are no duplicate records that could inflate model performance or skew analysis.

```python
df.duplicated().sum()
```

**Result:** No duplicate rows found. Each of the 15,000 records is unique.
